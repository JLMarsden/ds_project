# Logistic Regression Project 

In this project we will be working with a fake advertising data set, indicating whether or not a particular internet user clicked on an Advertisement. We will try to create a model that will predict whether or not they will click on an ad based off the features of that user.

This data set contains the following features:

* 'Daily Time Spent on Site': consumer time on site in minutes
* 'Age': cutomer age in years
* 'Area Income': Avg. Income of geographical area of consumer
* 'Daily Internet Usage': Avg. minutes a day consumer is on the internet
* 'Ad Topic Line': Headline of the advertisement
* 'City': City of consumer
* 'Male': Whether or not consumer was male
* 'Country': Country of consumer
* 'Timestamp': Time at which consumer clicked on Ad or closed window

# Logistic Regression

Now it's time to do a train test split, and train our model!

You'll have the freedom here to choose columns that you want to train on!

** Split the data into training set and testing set using train_test_split**

```python
from sklearn.model_selection import train_test_split
```

* Machine‑learning models require numeric features only, so passing this row directly into a model will cause errors such as “could not convert string to float” or shape mismatch.

- We need to drop data and convert what we reasonably can. 

```python
ad_data['Timestamp'] = pd.to_datetime(ad_data['Timestamp'])
ad_data['Hour'] = ad_data['Timestamp'].dt.hour
ad_data['Day'] = ad_data['Timestamp'].dt.day
ad_data['Month'] = ad_data['Timestamp'].dt.month
ad_data['Year'] = ad_data['Timestamp'].dt.year
ad_data = ad_data.drop(columns=['Timestamp'])

ad_data  = pd.get_dummies(ad_data , columns=['Ad Topic Line','City','Country'], drop_first=True)


# 1. Define X (everything except the 'Survived' column)
# 2. Define y (only the 'Survived' column)
X = ad_data.drop('Clicked on Ad', axis=1)  #not using inplace=True so train is just removing in memory Survived
y = ad_data['Clicked on Ad']

# 3. Perform the split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)
ad_data.head()
```

```python
from sklearn.linear_model import LogisticRegression
```

** Train and fit a logistic regression model on the training set.**

```python
logmodel = LogisticRegression(max_iter=1000)
y_train.tail()
```

```python
logmodel.fit(X_train,y_train)
```

## Predictions and Evaluations
** Now predict values for the testing data.**

```python
predictions = logmodel.predict(X_test)
```

** Create a classification report for the model.**

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,predictions))
```

## Create a matrix with a graph 
- This is the standard and most useful graph for classification model output.

Which one should you use?
- Confusion matrix → best for understanding mistakes
- Bar chart of precision/recall/F1 → best for comparing class performance
- ROC curve → best for evaluating threshold‑independent performance


```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,predictions)
```

```python
# y_test = true labels
# y_pred = model predictions

cm = confusion_matrix(y_test, predictions)

plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Pred 0', 'Pred 1'],
            yticklabels=['Actual 0', 'Actual 1'])

plt.title('Confusion Matrix')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```
