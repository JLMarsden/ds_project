```python
import os
from dotenv import load_dotenv
import kagglehub
import pandas as pd
```

```python
print(os.getenv("KAGGLE_USERNAME"))
print(os.getenv("KAGGLE_KEY"))
```

```python
# Load credentials from .env
load_dotenv()

# Set environment variables for authentication
os.environ["KAGGLE_USERNAME"] = os.getenv("KAGGLE_USERNAME")
os.environ["KAGGLE_KEY"] = os.getenv("KAGGLE_KEY")

# Download the latest version of a dataset (Example: 911 Emergency)
# Format: "owner/dataset-name"
path = kagglehub.dataset_download("mchirico/montcoalert")

print(f"Data downloaded to: {path}")

# Load the CSV into a DataFrame
# kagglehub returns the path to the folder; you may need to append the filename
df = pd.read_csv(f"{path}/911.csv")
```
