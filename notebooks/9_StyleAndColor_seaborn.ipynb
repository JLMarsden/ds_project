```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.set_style("whitegrid") # Set the style of the plot to whitegrid
sns.countplot(x="sex", data=tips, hue="smoker") # Create a count plot of the tips dataset
plt.show() # Show the plot
```

```python
#Use matplotlib to change the size of the plot on the fly over seaborn
plt.figure(figsize=(12,3)) # Set the figure size to 12 inches by 3 inches
sns.countplot(x="sex", data=tips, hue="smoker") # Create a count plot of the tips dataset
plt.show() # Show the plot
```
