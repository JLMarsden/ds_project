```python
%matplotlib inline
import seaborn as sns
import matplotlib.pyplot as plt


tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.lmplot(x="total_bill", y="tip", data=tips, hue="sex", markers=["o", "x"])
plt.show()
```

```python
# Simple linear regression model is built on matplotlib and seaborn. It is used to plot the data points and fit a linear regression model to the data. The x-axis 
# represents the total bill amount, while the y-axis represents the tip amount. The hue parameter is used to differentiate between
# we can add spome more parameters check the documentation for more details
sns.lmplot(x="total_bill", y="tip", data=tips, col="day", hue="sex", markers=["o", "x"], aspect=0.5,height=4, scatter_kws={"s": 100})
plt.show()  
```
