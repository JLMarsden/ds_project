```python
# 1. Create a grid of X and Y values
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
x_grid, y_grid = np.meshgrid(x, y)

# 2. Generate random "Z" data (a sin wave + random noise)
z = np.sin(np.sqrt(x_grid**2 + y_grid**2)) + np.random.normal(0, 0.1, x_grid.shape)

# 3. Create the Surface Plot
fig = go.Figure(data=[go.Surface(z=z, x=x, y=y)])

# 4. Style the layout
fig.update_layout(
    title='Random 3D Surface Plot',
    autosize=False,
    width=800,
    height=800,
    margin=dict(l=65, r=50, b=65, t=90),
    scene = dict(
        xaxis_title='X Axis',
        yaxis_title='Y Axis',
        zaxis_title='Z Axis'
    )
)

fig.show()
```
